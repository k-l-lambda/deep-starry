```python
from starry.transformer.layers import EncoderLayer


enc_starry = EncoderLayer(d_model=256, d_inner=1024, n_head=8, d_k=32, d_v=32)
enc_starry
```

```python
from transformers import LlamaConfig
from transformers.models.llama.modeling_llama import LlamaDecoderLayer


config = LlamaConfig(hidden_size=256, intermediate_size=1024, num_attention_heads=8)
enc_llama = LlamaDecoderLayer(config, 0)
enc_llama
```

```python
import torch


x = torch.randn(1, 4, 256)
x1 = enc_starry(x)[0]
x1.shape
```

```python
pos = torch.arange(4)[None]
x2 = enc_llama(x, position_ids=pos)
x2[0].shape
```
