# random single measure pair similarity

```python
ps = []

for i in range(1000):
	i1, i2 = torch.randint(summaries.shape[0], ()), torch.randint(summaries.shape[0], ())
	s1, s2 = summaries[i1], summaries[i2]
	product = (s1 * s2).sum(dim=-1) / (s1.norm(dim=-1) * s2.norm(dim=-1))
	ps.append(product)

ps = torch.stack(ps)
ps.shape
```

```python
ps, ps.abs().mean(), ps.max()
```
