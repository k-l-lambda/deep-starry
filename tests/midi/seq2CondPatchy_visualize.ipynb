# Seq2CondMidiPatchy — conditioned-MIDI measurewise (midiseq2) dataset visualization

Visualizes the joint-sequence dataset packed by `tools/midi/buildCondMidiseq2Measurewise.py`:
each song is ONE sequence `[lilylet score patches] ++ [midiseq2 midi patches]`, with a custom
windowed/block attention mask coupling midi measures to lilylet measures.

Unlike the event-patch `CondMidiPatchy` (see `condPatchy_visualize.ipynb`), the MIDI side here is
**midiseq2** (vocab 838): one MEASURE = a variable-length midiseq2 token run, chunked into
`patch_size`-wide patches, terminated by an `<eom>` **token** (id 4) — there is NO standalone
`<eom>` patch. Patches are int16 in a `patch_size=64` frame; lilylet ids live in columns `[0,16)`.

A short sample is the running example.

```python
from pathlib import Path
import sys, os

REPO_ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'starry').is_dir())
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

import torch
import numpy as np
import matplotlib.pyplot as plt

from starry.lilylet.data.patchifier import LilyletTokenizer
from starry.midi.data.seq2CondPachifier import Midiseq2Tokenizer, EOM_ID, PAD_ID, BOS_ID, EOS_ID
from starry.midi.data.seq2CondPatchy import Seq2CondMidiPatchy, _get_store, build_vis

DATASET_DIR = Path.home() / 'data' / 'lilylet' / 'midi-measurewise' / 'test20260629'
ROOT = str(DATASET_DIR / 'cond-midiseq2.lmmw.pt')
print('repo:', REPO_ROOT)
print('root:', ROOT)
```

## 1. Load the store and pick the shortest sample as the example

```python
store = _get_store(ROOT)
idx = store.artifact
print('format     :', idx['format'])
print('tokenizers :', idx['tokenizers'])
print('config     :', idx['config'])
print('items      :', len(store))

# shortest sample = clearest example
lengths = [(i, store.get(i)['patches'].shape[0]) for i in range(len(store))]
EX = min(lengths, key=lambda kv: kv[1])[0]
item = store.get(EX)
L = int(item['lyl_count'])
T = item['patches'].shape[0]
PS = item['patches'].shape[1]
print(f'\nexample #{EX}: id={item["id"]}')
print(f'  T={T}  patch_size={PS}  lyl_count L={L}  midi patches={T - L}')
print(f'  M_lyl={item["M_lyl"]}  M_midi={item["M_midi"]}  lyl_patch_size={item.get("lyl_patch_size")}')
```

## 2. Sequence layout: modality + measure indices

`modality` 0=lilylet / 1=midi; `measures` = each patch's own measure; `src_measures` =
lilylet-aligned measure (midi i -> source_measure). Measure boundaries show up as the `<eom>`
TOKEN inside each measure's LAST patch (not a separate patch), so we mark the last patch of each
midi measure as the bar boundary.

```python
tk = Midiseq2Tokenizer()
mod = item['modality'].numpy()
own = item['measures'].numpy()
src = item['src_measures'].numpy()
patches = item['patches'].long()

# a patch that CONTAINS the <eom> token is a measure-ending patch (last patch of its measure).
has_eom = [(int(item['modality'][i]) == 1 and (patches[i] == EOM_ID).any().item()) for i in range(T)]
eom_rows = [i for i, h in enumerate(has_eom) if h]
print('measure-ending midi patches (contain <eom> token):', eom_rows[:20], '...' if len(eom_rows) > 20 else '')
print('count:', len(eom_rows), '== M_midi:', item['M_midi'])

fig, ax = plt.subplots(3, 1, figsize=(13, 5), sharex=True)
ax[0].plot(mod, drawstyle='steps-post'); ax[0].set_ylabel('modality'); ax[0].set_yticks([0, 1]); ax[0].set_yticklabels(['lyl', 'midi'])
ax[0].axvline(L, color='r', ls='--', lw=1); ax[0].set_title(f'sequence layout (red = lyl/midi boundary at {L})')
ax[1].plot(own, drawstyle='steps-post', color='tab:green'); ax[1].set_ylabel('own measure')
ax[2].plot(src, drawstyle='steps-post', color='tab:orange'); ax[2].set_ylabel('src (lyl-aligned)')
for r in eom_rows:
    ax[1].axvline(r, color='gray', alpha=0.3, lw=0.5)
ax[2].set_xlabel('patch index')
plt.tight_layout(); plt.show()
```

## 3. Decode the patches back to text

lilylet patches (cols `[0,16)`) -> lilylet tokens; midiseq2 patches -> midiseq2 token strings
(`tk.tokens[id]`). The trailing `<eom>` marks each measure's end; `<pad>` fills the patch tail.

```python
lt = LilyletTokenizer(str(REPO_ROOT / 'assets' / 'lilylet-tokenizer.json'))
LPS = int(item.get('lyl_patch_size', 16))

def decode_lyl_patch(row):
    toks = [lt.text_by_id.get(int(t), '?') for t in row[:LPS] if int(t) not in (lt.pad_id, lt.bos_id, lt.eos_id)]
    return ''.join(toks)

def decode_seq2_patch(row):
    return ' '.join(tk.tokens[int(t)] for t in row if int(t) != PAD_ID)

print('=== lilylet segment (first 8 patches) ===')
for i in range(min(8, L)):
    print(f'  [{i:3d}] m{own[i]:>2}  {decode_lyl_patch(patches[i])!r}')

print('\n=== midiseq2 segment: first full measure (measure 1), patch by patch ===')
m1_rows = [i for i in range(L, T) if own[i] == 1]
for i in m1_rows:
    print(f'  [{i:4d}] m{own[i]:>2}  {decode_seq2_patch(patches[i])}')
```

## 4. The custom attention mask

Built by `build_vis` (reused verbatim from `condPatchy`): lyl->lyl full causal, lyl->midi
forbidden, midi->lyl cross block (`j<=i<j+W_cross`, lyl header globally visible), midi->midi
windowed-causal by measure (`[m-W_midi+1, m]`). The midiseq2 measure tags are identical to the
event-patch ones, so the mask logic is unchanged. True = query (row) may attend key (col).

```python
W_MIDI, W_CROSS = 2, 2
vis = build_vis(item['modality'].long(), item['measures'].long(), item['src_measures'].long(), W_MIDI, W_CROSS)
print('vis shape:', tuple(vis.shape), '| density: %.3f' % vis.float().mean().item())

fig, ax = plt.subplots(figsize=(8, 8))
ax.imshow(vis.numpy(), cmap='Greys', interpolation='nearest', aspect='equal')
ax.axvline(L - 0.5, color='r', ls='--', lw=1); ax.axhline(L - 0.5, color='r', ls='--', lw=1)
ax.set_xlabel('key (attended)'); ax.set_ylabel('query'); ax.set_title(f'attention mask (W_midi={W_MIDI}, W_cross={W_CROSS})')
ax.text(L * 0.5, L * 0.5, 'lyl->lyl', color='tab:blue', ha='center', fontsize=11)
ax.text(L * 0.5, L + (T - L) * 0.5, 'midi->lyl\n(cross)', color='tab:orange', ha='center', fontsize=10)
ax.text(L + (T - L) * 0.5, L + (T - L) * 0.5, 'midi->midi\n(window)', color='tab:green', ha='center', fontsize=10)
plt.tight_layout(); plt.show()
```

## 5. Mask sanity checks

Confirm the four block rules and that no real query row is empty (SDPA safety).

```python
v = vis
checks = {
    'no future leak (upper triangle empty)': not v.triu(1).any().item(),
    'lyl->lyl is full lower-triangular':   bool((v[:L, :L] == torch.tril(torch.ones(L, L, dtype=torch.bool))).all()),
    'lyl->midi all forbidden':             not v[:L, L:].any().item(),
    'every query row non-empty':             bool((v.sum(1) >= 1).all()),
}
for k, ok in checks.items():
    print(('  OK  ' if ok else ' FAIL ') + k)
assert all(checks.values())

# midi->midi window width per query, EXCLUDING the always-visible measure-0 header
own_t = item['measures'].long()
spans = []
for q in range(L, T):
    ks = v[q, L:].nonzero().flatten()
    ms = own_t[L + ks]
    ms = ms[ms > 0]                      # drop the midi header (measure 0, globally visible)
    if len(ms):
        spans.append(int(ms.max() - ms.min()) + 1)
print('\nmidi->midi attended measure-span (excl. header): max=%d (<= W_midi=%d)' % (max(spans), W_MIDI))
assert max(spans) <= W_MIDI
```

## 6. Cross-coupling zoom: which lilylet measures does each midi measure see?

For each midi measure i, the visible lilylet measures should be `{0 (header)} + [i-W_cross+1 .. i]`.
We use each measure's LAST patch (the one carrying its `<eom>` token) as the representative row.

```python
rows = []
for r in eom_rows:
    i = int(own[r])
    vis_lyl_measures = sorted(set(int(own[k]) for k in v[r, :L].nonzero().flatten().tolist()))
    rows.append((i, vis_lyl_measures))
for i, vm in rows[:12]:
    print(f'  midi measure {i:>2} (<eom> patch) sees lyl measures: {vm}')
```

## 7. Through the feeder: a collated batch

Confirm `collateBatch` produces the 4D bool mask, supervision targets on midi only, padding
handled (two samples of different length), and the int16-safe token ids / `lyl_patch_size`.

```python
ds = Seq2CondMidiPatchy(ROOT, '0..19/20', device='cpu', shuffle=False, w_midi=W_MIDI, w_cross=W_CROSS)
order = sorted(range(len(store)), key=lambda i: store.get(i)['patches'].shape[0])
batch = ds.collateBatch([ds._item(order[0]), ds._item(order[3])])
for k, val in batch.items():
    if torch.is_tensor(val):
        print(f'  {k:16} {tuple(val.shape)} {val.dtype}')
    else:
        print(f'  {k:16} {val}')

tgt = batch['input_targets']; modb = batch['modality']
print('\ntargets only on midi segment:', bool(((tgt == 1) & (modb == 0)).sum() == 0))
print('target count per sample      :', tgt.sum(1).tolist())
print('attn_mask is 4D bool         :', batch['attn_mask'].dim() == 4 and batch['attn_mask'].dtype == torch.bool)
print('max midiseq2 token id        :', int(batch['input_patches'].max()), '(vocab', tk.vocab_size, ')')
```
