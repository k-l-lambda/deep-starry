```python
import os


with open(os.path.expanduser('~/data/maestro-v3.0.0/midiseq.yaml')) as f:
	text = f.read()
len(text), text[:100]
```

```python
import re


words = re.findall(r'\b[A-Z][a-z]+\d+\b', text)
len(words), words[:100]
```

```python
n_total = len(words)

n_on = len([w for w in words if w.startswith('On')])
n_off = len([w for w in words if w.startswith('Off')])
n_vel = len([w for w in words if w.startswith('Vel')])
n_shift = len([w for w in words if w.startswith('Shift')])
n_pedal = len([w for w in words if w.startswith('Sustain') or w.startswith('Soft') or w.startswith('Sostenuto')])
n_sustain = len([w for w in words if w.startswith('Sustain')])
n_soft = len([w for w in words if w.startswith('Soft')])
n_soft = len([w for w in words if w.startswith('Soft')])
n_sostenuto = len([w for w in words if w.startswith('Sostenuto')])

dict(n_on=n_on, n_off=n_off, n_vel=n_vel, n_shift=n_shift, n_pedal=n_pedal, n_sustain=n_sustain, n_soft=n_soft, n_sostenuto=n_sostenuto), dict(r_on=n_on/n_total, r_off=n_off/n_total, r_vel=n_vel/n_total, r_shift=n_shift/n_total, r_pedal=n_pedal/n_total, r_sustain=n_sustain/n_total, r_soft=n_soft/n_total)
```
