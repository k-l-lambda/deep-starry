```python
import os
import yaml


scores = yaml.safe_load(open(os.path.expanduser('~/data/paraff/ly.yaml'), 'r'))
scores
```

```python
import os

from starry.utils.config import Configuration


TRAINING_DIR = os.environ.get('TRAINING_DIR')

config = Configuration.createOrLoad(os.path.join(TRAINING_DIR, 'paraff/20241113-paraff-vae-shared-241104-d256-l4/'))
config.data
```

```python
tokens = config['_vocab'].split(',')
tokens
```

```python
T2I = dict([(t, i) for i, t in enumerate(tokens)])
T2I
```

```python
s0 = scores['BWV-114']
s0 = [m for m in s0.values()]
s0 = ['BOM' + m.split('BOM', 1)[-1] for m in s0]
s0
```

```python
m_ids = [list(map(lambda x: T2I[x], m.replace('\n', ' ').split(' '))) for m in s0]
m_ids
```

```python
import torch


model = torch.jit.load(os.path.join(TRAINING_DIR, 'paraff/20241113-paraff-vae-shared-241104-d256-l4/model_42851_loss_1.930e-02.pt'))
model.eval()
model
```

```python
msums = []

for ids in m_ids:
	with torch.no_grad():
		input_ids = torch.zeros(256).long()
		input_ids[:len(ids)] = torch.as_tensor(ids).long()
		s = model(input_ids[None], torch.zeros(1))
	s /= s.norm(dim=-1)
	msums.append(s)

msums = torch.cat(msums, dim=0)
msums.shape
```

```python
from tqdm import tqdm


sum_lib = {}
for name, score in tqdm(scores.items()):
	measures = [m for m in score.values()][:16]
	measures = ['BOM' + m.split('BOM', 1)[-1] for m in measures]
	measure_ids = [list(map(lambda x: T2I[x], m.replace('\n', ' ').split(' '))) for m in measures]

	msums = []
	for ids in measure_ids:
		with torch.no_grad():
			input_ids = torch.zeros(256).long()
			input_ids[:len(ids)] = torch.as_tensor(ids).long()
			s = model(input_ids[None], torch.zeros(1))
		msums.append(s)

	msums = torch.cat(msums, dim=0)
	sum_lib[name] = msums

sum_lib
```
