```python
from starry.paraff.midiseq import T2I


with open('../../intelli-piano/footages/40318-p0.spartito.midiseq.txt') as file:
	segments = file.readlines()
	segments = [[T2I[token] for token in segment.strip().split(' ')] for segment in segments]

segments
```

```python
import os

from starry.utils.model_factory import loadModel
from starry.utils.config import Configuration


TRAINING_DIR = os.environ.get('TRAINING_DIR')

config = Configuration.createOrLoad(os.path.join(TRAINING_DIR, 'paraff/20241130-midiseq-paraff-se-ly+cc-d256-l8-dw0.06'))
model = loadModel(config['model'], postfix='JitEnc')
model
```

```python
import torch


checkpoint = torch.load(config.localPath(config['best']), map_location='cpu')
checkpoint
```

```python
model.load_state_dict(checkpoint['model'], strict=False)
```

```python
input = torch.tensor([2] + segments[0] + [3]).long()[None]
input
```

```python
model.eval()
```
